```python
import pandas as pd

col_list = ["id","first","last","gender","Marks","selected"]

df = pd.read_csv("sample.csv", usecols=col_list)

print(df)

mean1 = df['Marks'].mean()
sum1 = df['Marks'].sum()
max1 = df['Marks'].max()
min1 = df['Marks'].min()
count1 = df['Marks'].count()
median1 = df['Marks'].median()
sd1 = df['Marks'].std()
var1 = df['Marks'].var()

print("\nMean Marks =", mean1)
print("Sum of Marks =", sum1)
print("Maximum Marks =", max1)
print("Minimum Marks =", min1)
print("Count =", count1)
print("Median =", median1)
print("Standard Deviation =", sd1)
print("Variance =", var1)

print("\nEnd of Summary")
```

```python
class Student:

    def __init__(self, regno, name):
        self.regno = regno
        self.name = name

    def display(self):
        print("Register No:", self.regno)
        print("Name:", self.name)

s1 = Student(101, "Sam")

s1.display()
```
