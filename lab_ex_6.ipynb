```python
#logistic regression
import pandas as pd
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

#create dataset
x,y=make_blobs(n_samples=200,centers=3,n_features=3,random_state=0)

#create Dataframe
df=pd.DataFrame(dict(x=x[:,0],y=x[:,1],label=y))

#print first five records
print("Top five records\n")
print(df.head(5))

#split data
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.40,random_state=0)

#create model
model=LogisticRegression()

#Train Model
model.fit(x_train,y_train)

#predict
predicted=model.predict(x_test)

#classification report
print("\n classification report")
print(classification_report(y_test,predicted))
```

```python
#L1 regularization-lasso
import numpy as np
from sklearn.linear_model import Lasso
#ctreate dataset
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([2,4,6,8,10])

#create lasso model
model=Lasso(alpha=0.1)

#train model
model.fit(x,y)
#display result
print("L1 regression_lasoo")
print("Coefficient:",model.coef_)
print("Intercept:",model.intercept_)
```

```python
#L2 regularisatin -Ridge
import numpy as np
from sklearn.linear_model import Ridge

#create dataset
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([2,4,6,8,10])

#create ridge model
model=Ridge(alpha=0.1)

#train model
model.fit(x,y)

#display result
print("L2 regularixation_Ridge")
print("Coefficient:",model.coef_)
print("Intercept:",model.intercept_)
```

```python
#elastic net regression
import numpy as np
from sklearn.linear_model import ElasticNet

#create dataset
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([2,4,6,8,10])

#create elastic net model
model=ElasticNet(alpha=0.1,l1_ratio=0.5)

#train model
model.fit(x,y)

#display result
print("Elastic net regression")
print("Coefficient:",model.coef_)
print("Intercept:",model.intercept_)
```

```python
#MSE 
import numpy as np
from sklearn.metrics import mean_squared_error

# Actual values
y_actual = np.array([3, 5, 7, 9])

# Predicted values
y_predicted = np.array([2, 5, 8, 10])

# Calculate MSE
mse = mean_squared_error(y_actual, y_predicted)

print("Mean Squared Error:", mse)
```

```python
#MAE
import numpy as np
from sklearn.metrics import mean_absolute_error

# Actual values
y_actual = np.array([3, 5, 7, 9])

# Predicted values
y_predicted = np.array([2, 5, 8, 10])

# Calculate MAE
mae = mean_absolute_error(y_actual, y_predicted)

print("Mean Absolute Error:", mae)
```
